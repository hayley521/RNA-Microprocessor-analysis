```r
df <- read.table("/wanglab1/gangx/data/LYW/microRNA-seq/240704_hairpin-miRNA.csv",sep=',',header=T,row.names=1)
```

```r
df<-df[rowMeans(df)>1,]
```

```r
asint <- function(x) {  
  as.integer(round(x,digits=0))}
```

```r
df_raw<-as.data.frame(lapply(df,asint),row.names=row.names(df))
```

```r
colnames(df_raw)
```

```r
colData=data.frame(id = c('R0_1','R0_2','R4_1','R4_2','R8_1','R8_2','R12_1','R12_2','R24_1','R24_2','R72_1','R72_2'),
    group = c('R0','R0','R4','R4','R8','R8','R12','R12','R24','R24','R72','R72'))
```

```r
unique(colData$group)
```

```r
colData$group <- factor(colData$group, levels = c('R0','R4','R8','R12','R24','R72'))
```

```r
dds <- DESeqDataSetFromMatrix(countData = df_raw, colData = colData, design = ~ group)
```

```r
dds <- estimateSizeFactors(dds, controlGenes=1:2 )
```

```r
dds_norm <- DESeq(dds) 
```

```r
normalized_counts <- counts(dds_norm,normalized=T) 
```

```r
write.csv(normalized_counts,"result/hairpin_normalized_counts.csv")
```

```r
function_anlysis <- function(DEseq2,sample_name) {
    DEseq2 <- DEseq2 %>% filter(gene_type=="protein_coding")
    DEseq2$ENSEMBL<-str_split(DEseq2$Row.names,"\\.",n=2,simplify = TRUE)[,1]
    gene.df <- bitr(DEseq2$ENSEMBL,
               fromType = "ENSEMBL",
               toType = c("SYMBOL","ENTREZID"),
               OrgDb = org.Mm.eg.db)
    DEseq2_merge=merge(DEseq2,gene.df,by.y="ENSEMBL",by.x="ENSEMBL")
    genelist <- DEseq2_merge$log2FoldChange
    names(genelist)<-DEseq2_merge$ENTREZID
    genelist <- sort(genelist, decreasing = TRUE)
    gsea <- gseGO(
        geneList = genelist, 
        OrgDb   = org.Mm.eg.db,
        ont = "BP", eps=0, 
        minGSSize = 20,
        maxGSSize = 500,
        pvalueCutoff = 0.05,
        pAdjustMethod  = "BH")
    write.csv(gsea,file=paste("result2-1/GSEA/",sample_name,"_GO_GSEA.csv", sep = ""), quote=T, sep=",", row.names=F)
    kk <- gseKEGG(
        geneList  = genelist,
        keyType  = 'kegg',
        organism = 'mmu',
        eps=0,
        minGSSize = 20,
        maxGSSize = 500,
        pvalueCutoff = 0.05,
        pAdjustMethod  = "BH"
    )
    write.csv(kk,file=paste("result2-1/KEGG/",sample_name,"_KEGG_GSEA.csv", sep = ""), quote=T, sep=",", row.names=F)
    ego_up <- enrichGO(
        gene = (DEseq2 %>% filter(log2FoldChange>  0.75, pvalue<0.05 ))$ENSEMBL,
        OrgDb = org.Mm.eg.db, 
        keyType = 'ENSEMBL', 
        ont = "BP", 
        pvalueCutoff = 0.05,
        readable = T)
    ego_down <- enrichGO(
        gene = (DEseq2 %>% filter(log2FoldChange< -0.75, pvalue<0.05 ))$ENSEMBL,
        OrgDb = org.Mm.eg.db, 
        keyType = 'ENSEMBL', 
        ont = "BP", 
        pvalueCutoff = 0.05,
        readable = T)
    write.csv(simplify(ego_up,cutoff=0.7, by="p.adjust", select_fun=min),  file=paste("result2-1/GO/",sample_name,"_upGO.csv", sep = ""), quote=T, sep=",", row.names=F)
    write.csv(simplify(ego_down,cutoff=0.7, by="p.adjust", select_fun=min),file=paste("result2-1/GO/",sample_name,"_downGO.csv", sep = ""), quote=T, sep=",", row.names=F)
}
```
